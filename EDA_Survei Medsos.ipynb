```python
import pandas as pd # Library untuk manipulasi data
import matplotlib.pyplot as plt # Library untuk visualisasi
import seaborn as sns # Library untuk visualisasi

# Mengatur gaya visual plot agar lebih menarik
sns.set_style("whitegrid")
```

# **3. Membaca Dataset**

```python
# Membaca dataset survei medsos
df = pd.read_csv("Survei_Medsos_Bersih.csv")  # ganti dengan path dataset kamu
df.head()
```

# **5. Analisis Frekuensi**

Pertanyaan analitis:
1. Seberepa banyak jumlah responden berdasarkan jenis kelamin?
2. Apa pekerjaan dominan responden dalam survei ini?
2. Platform media sosial apa yang paling sering digunakan?

```python
## 1. Jenis Kelamin & Pekerjaan
plt.figure(figsize=(10,4))

# Subplot 1: Distribusi Jenis Kelamin
plt.subplot(1,2,1)
sns.countplot(
    data=df,
    x='Jenis_Kelamin',
    order=df['Jenis_Kelamin'].value_counts().index,
    palette={'Perempuan': 'orange', 'Laki-Laki': 'blue'}  # sesuaikan label jika berbeda
)
plt.title('Distribusi Jenis Kelamin (Bar chart)')
```

Bar chart ini menggambarkan frekuensi responden berdasarkan jenis kelamin. Dari grafik ini dapat terlihat perbandingan jumlah responden laki-laki dan perempuan. Semakin tinggi batang, semakin banyak responden dalam kategori tersebut, dan terlihat bahwa perempuan sedikit lebih dominan dibanding laki-laki.

Interpretasi Hasil :

Visualisasi menunjukkan bahwa responden didominasi oleh perempuan dengan jumlah sedikit lebih banyak dibandingkan laki-laki. Hal ini mengindikasikan bahwa dalam survei ini, partisipasi perempuan lebih tinggi. Kondisi ini bisa memengaruhi hasil analisis selanjutnya, karena persepsi dan dampak penggunaan media sosial mungkin berbeda antar gender.

```python
# Subplot 2: Top 8 Pekerjaan
plt.subplot(1,2,2)
top_jobs = df['Pekerjaan'].value_counts().nlargest(8).index
sns.countplot(data=df, y='Pekerjaan', order=top_jobs)
plt.title('Top 8 Pekerjaan (Bar chart)')

plt.tight_layout()
plt.show()
```

Bar chart ini menggambarkan frekuensi responden berdasarkan pekerjaan, khususnya delapan kategori pekerjaan dengan jumlah responden terbanyak. Dari grafik ini dapat diketahui bahwa kategori pelajar/mahasiswa merupakan kelompok paling dominan dengan jumlah responden jauh lebih tinggi dibandingkan kategori pekerjaan lainnya. Semakin panjang batang, semakin banyak responden dalam kategori pekerjaan tersebut.

Interpretasi Hasil :

Mayoritas responden adalah pelajar/mahasiswa dengan jumlah jauh lebih tinggi dibanding kategori pekerjaan lainnya. Hal ini menunjukkan bahwa survei ini lebih banyak mencerminkan pola penggunaan media sosial di kalangan generasi muda atau mereka yang masih dalam pendidikan formal. Dampak media sosial pada aspek produktivitas dan kontrol diri kemungkinan besar lebih relevan pada kelompok ini dibanding kelompok pekerjaan lain yang jumlahnya lebih sedikit.

```python
platform_counts = df['Sering'].value_counts()
platform_counts

```

```python
print("\nVisualisasi: Bar Chart")
plt.figure(figsize=(10,6))
sns.countplot(x="Sering", data=df, order=platform_counts.index)
plt.title("Jumlah Penggunaan Platform Media Sosial (Bar Chart)")
plt.xlabel("Platform")
plt.ylabel("Jumlah Responden")
plt.show()

```


Bar chart ini menggambarkan frekuensi jawaban responden berdasarkan kategori tertentu. Dari grafik ini kita dapat melihat kategori mana yang paling dominan dipilih responden. Semakin tinggi batang, semakin banyak responden yang memilih kategori tersebut.

Interpretasi hasil:

Instagram dan WhatsApp mendominasi penggunaan, sedangkan platform lain lebih jarang dipilih.

# **6. Analisis Distribusi**

1. Pertanyaan Analitis: Bagaimana sebaran waktu harian penggunaan media sosial?
2. Pertanyaan Analitis : Bagaimana distribusi skor gangguan produktivitas akibat penggunaan media sosial pada responden survei?

```python
print("\nVisualisasi: Bar Chart")
plt.figure(figsize=(10,6))
sns.countplot(x="Waktu_Harian", data=df, order=df["Waktu_Harian"].value_counts().index)
plt.title("Distribusi Waktu Harian Penggunaan Media Sosial (Bar Chart)")
plt.xlabel("Durasi Harian")
plt.ylabel("Jumlah Responden")
plt.show()
```


Bar chart ini menggambarkan jawaban responden berdasarkan kategori tertentu. Dari grafik ini kita dapat melihat kategori mana yang paling dominan dipilih responden. Semakin tinggi batang, semakin banyak responden yang memilih kategori tersebut.

Interpretasi hasil:

Sebagian besar responden menggunakan media sosial lebih dari 2–3 jam per hari.

```python
print("\nVisualisasi: Histogram")
plt.figure(figsize=(10,6))
sns.histplot(df["Mengganggu_Produktivitas"], bins=5, kde=True)
plt.title("Distribusi Skor Gangguan Produktivitas (Histogram)")
plt.xlabel("Skor")
plt.ylabel("Frekuensi")
plt.show()
```

Histogram ini menunjukkan bagaimana sebaran skor responden pada variabel terkait. Dari pola distribusi, kita bisa melihat kecenderungan responden memberikan skor tertentu. Jika mayoritas skor berada pada nilai tinggi, maka responden cenderung setuju/merasakan dampak tersebut. Sebaliknya, jika banyak skor rendah, berarti mayoritas responden kurang merasakan dampaknya.

Interpretasi hasil:

Histogram menunjukkan bahwa mayoritas responden memberikan skor 3 dan 4 terhadap pertanyaan seberapa besar media sosial mengganggu produktivitas mereka. Hal ini mengindikasikan bahwa sebagian besar responden merasa penggunaan media sosial cukup hingga sering menurunkan produktivitas. Sebaliknya, hanya sedikit responden yang memberikan skor 1 (tidak terganggu) maupun skor 5 (sangat terganggu), sehingga dapat disimpulkan bahwa tingkat gangguan yang dirasakan cenderung berada pada kategori sedang hingga tinggi, bukan pada kategori ekstrem.

# **7. Analisis Hubungan**

1. Apakah durasi harian penggunaan media sosial berhubungan dengan tingkat Tidak Sadar Waktu dan Tidak_Bisa_Kontrol_Diri?
2. Pertanyaan Analitis: Apakah durasi harian penggunaan media sosial berhubungan dengan tingkat FOMO?

```python
plt.figure(figsize=(10,6))
df_melt = df.melt(id_vars="Waktu_Harian",
                  value_vars=["Tidak_Sadar_Waktu", "Tidak_Bisa_Kontrol_Diri"],
                  var_name="Aspek", value_name="Skor")
sns.barplot(x="Waktu_Harian", y="Skor", hue="Aspek", data=df_melt, ci="sd")
plt.title("Perbandingan Skor Berdasarkan Waktu Harian (Grouped Barplot)")
plt.xlabel("Waktu Harian (jam)")
plt.ylabel("Skor")
plt.show()
```

Barplot ini menunjukkan rata-rata skor Tidak Sadar Waktu dan Tidak Bisa Kontrol Diri pada tiap kategori durasi harian penggunaan media sosial, lengkap dengan variasinya melalui error bar. Dari barplot ini kita bisa membandingkan kecenderungan antar kategori durasi, serta melihat apakah ada perbedaan pola yang berarti di antara kelompok pengguna dengan waktu harian yang berbeda.

Interpretasi Hasil :

barplot menunjukkan bahwa semakin lama durasi harian penggunaan media sosial, rata-rata skor Tidak Sadar Waktu dan Tidak Bisa Kontrol Diri cenderung lebih tinggi, meski perbedaannya tidak terlalu besar. Ini mengindikasikan adanya pengaruh durasi, tetapi faktor lain juga kemungkinan berperan.

```python
print("\nVisualisasi: Boxplot")
plt.figure(figsize=(10,6))
sns.boxplot(x="Waktu_Harian", y="Fomo", data=df)
plt.title("Hubungan Durasi Harian dengan Tingkat FOMO (Boxplot)")
plt.xlabel("Durasi Harian")
plt.ylabel("Skor FOMO")
plt.show()

```


Boxplot ini menunjukkan persebaran data berdasarkan kuartil, median, serta potensi outlier. Dari boxplot ini kita bisa membandingkan distribusi antar kategori, dan melihat apakah ada perbedaan signifikan di antara kelompok tersebut.

Interpretasi hasil:

Responden dengan durasi penggunaan lebih lama cenderung memiliki skor FOMO lebih tinggi.

# **8. Analisis Perbandingan**

Pertanyaan Analitis: Apakah ada perbedaan durasi penggunaan antara laki-laki dan perempuan?

```python
print("\nVisualisasi: Countplot (Bar Chart dengan Hue)")
plt.figure(figsize=(10,6))
sns.countplot(x="Waktu_Harian", hue="Jenis_Kelamin", data=df,
              order=df["Waktu_Harian"].value_counts().index)
plt.title("Perbandingan Waktu Harian Penggunaan Berdasarkan Jenis Kelamin (Bar Chart)")
plt.xlabel("Durasi Harian")
plt.ylabel("Jumlah Responden")
plt.show()

```


Bar chart ini menggambarkan jawaban responden berdasarkan kategori tertentu. Dari grafik ini kita dapat melihat kategori mana yang paling dominan dipilih responden. Semakin tinggi batang, semakin banyak responden yang memilih kategori tersebut.

Interpretasi hasil:

Baik laki-laki maupun perempuan menunjukkan pola penggunaan harian yang mirip, meski perempuan sedikit lebih banyak pada kategori >3 jam.

# **9. Hipotesis**

- Hipotesis 1 (Bar Chart – Distribusi Jenis Kelamin):
Partisipasi responden perempuan sedikit lebih tinggi daripada laki-laki. Hal ini dapat mengindikasikan bahwa topik kecanduan media sosial dan dampaknya mungkin lebih relevan atau menjadi perhatian yang lebih besar bagi perempuan dalam sampel survei ini.

- Hipotesis 2 (Bar Chart – Top 8 Pekerjaan):
Mayoritas responden adalah pelajar/mahasiswa, yang menunjukkan bahwa dampak negatif penggunaan media sosial terhadap produktivitas dan kontrol diri kemungkinan besar akan lebih terlihat pada kelompok usia muda yang aktivitas utamanya adalah belajar.

- Hipotesis 3 (Bar Chart – Jumlah Penggunaan Platform Media Sosial):
Instagram dan WhatsApp adalah platform yang paling sering digunakan, menunjukkan bahwa media sosial berbasis visual dan komunikasi personal menjadi pendorong utama interaksi digital yang berpotensi menyebabkan lupa waktu dan kecanduan.

- Hipotesis 4 (Bar Chart – Distribusi Waktu Harian):
Sebagian besar responden menggunakan media sosial lebih dari 2 jam setiap hari, yang mengindikasikan tingkat keterlibatan yang tinggi dan berisiko besar menimbulkan dampak negatif seperti gangguan produktivitas dan kesulitan mengontrol diri.

- Hipotesis 5 (Histogram – Distribusi Skor Gangguan Produktivitas):
Sebaran skor gangguan produktivitas terpusat pada nilai 3 dan 4, menunjukkan bahwa mayoritas responden secara sadar mengakui bahwa penggunaan media sosial cukup sering mengganggu efektivitas mereka dalam bekerja atau belajar.

- Hipotesis 6 (Grouped Barplot – Perbandingan Skor Berdasarkan Waktu Harian):
Semakin lama durasi penggunaan harian, semakin tinggi pula skor rata-rata untuk "Tidak Sadar Waktu" dan "Tidak Bisa Kontrol Diri", yang memperkuat dugaan adanya hubungan positif antara waktu penggunaan dan hilangnya kontrol serta kesadaran waktu.

- Hipotesis 7 (Boxplot – Hubungan Durasi Harian dengan Tingkat FOMO):
Responden dengan durasi penggunaan harian yang lebih panjang menunjukkan skor FOMO yang lebih tinggi, mengindikasikan bahwa rasa takut ketinggalan informasi atau tren menjadi salah satu pendorong utama untuk terus-menerus terhubung dengan media sosial.

- Hipotesis 8 (Bar Chart – Perbandingan Waktu Harian Berdasarkan Jenis Kelamin):
Meskipun perempuan sedikit lebih dominan dalam kategori penggunaan di atas 3 jam, pola penggunaan harian antara laki-laki dan perempuan secara umum tidak menunjukkan perbedaan yang drastis, ini menandakan bahwa durasi penggunaan media sosial yang tinggi adalah masalah yang umum terjadi di kedua gender.
