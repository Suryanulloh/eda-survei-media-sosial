# **2. Temukan penyakit data (Outliers, Missing values, Duplikasi, Inkonsistensi/Tipe data salah)**

**2.1 Cek Penyakit Data**

```python
## 4.1 Cek Missing Values
print("\nJumlah Missing Values per Kolom:")
print(df.isnull().sum())

## 4.2 Cek Duplikasi Data
print("\nJumlah Data Duplikat:", df.duplicated().sum())

## 4.3 Cek Inkonsistensi / Nilai Unik
print("\nCek Nilai Unik pada Kolom Kategorikal:")
for col in df.select_dtypes(include='object').columns:
    print(f"{col} -> {df[col].unique()}")

## 4.4 Cek Outliers (hanya kolom numerik)
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = df.select_dtypes(include=np.number).columns
for col in numeric_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot {col}")
    plt.show()
```

Kesimpulan Boxplot (Deteksi Penyakit Data)

A. Outliers

Terlihat adanya outlier pada beberapa variabel:

Mengganggu Produktivitas, Membuang Waktu, Tidak Sadar Waktu → outlier di sisi bawah (nilai 1).

Terpikirkan → outlier di sisi atas (nilai 5).

Outlier ini mengindikasikan adanya responden yang memberikan jawaban ekstrem dibanding mayoritas. Hal ini tidak selalu kesalahan data, bisa juga mencerminkan kondisi nyata, sehingga perlu dianalisis lebih lanjut sebelum diputuskan untuk dihapus atau dipertahankan.

B. Missing Values

Dari boxplot, tidak terlihat adanya missing values, karena semua variabel tetap memiliki distribusi yang utuh (box terbentuk sempurna tanpa celah).

C. Duplikasi

Boxplot tidak bisa langsung menunjukkan duplikasi data, sehingga aspek ini perlu diperiksa dengan metode lain (misalnya drop_duplicates).

D. Inkonsistensi / Tipe Data Salah

Semua variabel terbaca dengan baik dan konsisten dalam skala numerik (1–5). Tidak terlihat adanya nilai di luar rentang (misalnya 0 atau >5). Artinya tidak ada kesalahan tipe data atau inkonsistensi nilai yang terdeteksi melalui boxplot.

# **3. Tangani dan beri justifikasi**

**3.1 Menangani Penyakit Data**

```python
## 5.1 Tangani Missing Values
# Contoh: jika < 5%, hapus; jika lebih banyak, gunakan imputasi
df_clean = df.copy()
for col in df_clean.columns:
    if df_clean[col].isnull().sum() > 0:
        if df_clean[col].isnull().mean() < 0.05:
            df_clean = df_clean.dropna(subset=[col])
        else:
            if df_clean[col].dtype == 'object':
                df_clean[col].fillna(df_clean[col].mode()[0], inplace=True)
            else:
                df_clean[col].fillna(df_clean[col].median(), inplace=True)

## 5.2 Hapus Duplikasi
df_clean = df_clean.drop_duplicates()

## 5.3 Standarisasi Nilai (contoh untuk kolom Jenis_Kelamin)
df_clean['Jenis_Kelamin'] = df_clean['Jenis_Kelamin'].replace({
    'Laki-laki': 'Laki-Laki',
    'Perempuan': 'Perempuan'
})
```

**3.2 Simpan Dataset Bersih**

```python
df_clean.to_csv("Survei_Medsos_Bersih.csv", index=False)

print("\nDataset bersih berhasil disimpan: Survei_Medsos_Bersih.csv")
```
